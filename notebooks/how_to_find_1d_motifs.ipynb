# How do I find the most similar pair of time series sub-sequences (motifs)?

*Reproducing Slide 3 from Eamonn Keogh's "100 Time Series Data Mining Questions (with Answers)" using STUMPY.*

```python
import matplotlib.pyplot as plt
import stumpy
import pandas as pd

# 1. Load readily available dataset
df = pd.read_csv("https://raw.githubusercontent.com/TDAmeritrade/stumpy/main/docs/TCB_daily.csv")
m = 50  # Window size

# 2. Compute Matrix Profile in 1 line
mp = stumpy.stump(df["close"], m=m)

# 3. Find top motif pair
motif_idx = mp[:, 0].argmin()
nearest_neighbor_idx = mp[motif_idx, 1]

# 4. Plot time series and matrix profile
fig, axs = plt.subplots(2, 1, sharex=True, figsize=(10, 6))
axs[0].plot(df["close"], color="black")
axs[0].set_ylabel("Time Series")
axs[1].plot(mp[:, 0], color="tab:blue")
axs[1].axvline(x=motif_idx, color="tab:red", linestyle="--")
axs[1].axvline(x=nearest_neighbor_idx, color="tab:red", linestyle="--")
axs[1].set_ylabel("Matrix Profile")
plt.show()
```

### Summary
The two red dashed lines highlight the starting positions of the most similar pair of sub-sequences (motifs) of length $m = 50$.
