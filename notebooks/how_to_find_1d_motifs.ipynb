# How do I find the most similar pair of time series sub-sequences (motifs)?

**Reference:** Based on Question 1 from Eamonn Keogh's *100 Time Series Data Mining Questions (with Answers)*.

**Goal:** Given a single 1D time series, find the two sub-sequences of window size $m$ that are most similar to each other (the top motif pair).

```python
import numpy as np
import stumpy

# 1. Create or load a 1D time series (array-like, float64)
time_series = np.array([1.0, 1.3, 1.2, 5.0, 1.1, 1.3, 1.2, 0.9, 4.8], dtype=np.float64)
m = 3  # Window size (length of the sub-sequences to compare)

# 2. Compute the matrix profile (mp) and matrix profile indices
mp = stumpy.stump(time_series, m)

# 3. Extract the nearest neighbor pair (top motif)
motif_idx_1 = np.argmin(mp[:, 0])
motif_idx_2 = mp[motif_idx_1, 1]

print(f"Top motif found at index {motif_idx_1} and index {motif_idx_2}")
```

### How It Works

* **`mp[:, 0]`**: The Matrix Profile distances. The minimum value represents the most similar sub-sequence pair.
* **`mp[:, 1]`**: The Matrix Profile indices. Points directly to the starting index of the nearest neighbor.
